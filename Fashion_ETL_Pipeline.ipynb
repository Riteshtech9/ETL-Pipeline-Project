# 🧵 Fashion Dataset ETL Pipeline
This notebook performs an ETL (Extract, Transform, Load) process on the uploaded fashion dataset.

## ✅ Step 1: Import Libraries

```
import pandas as pd
import numpy as np
```

## ✅ Step 2: Extract - Load the Dataset

```
# Load the Fashion dataset (make sure the CSV is in the same directory)
df = pd.read_csv("e1a2fef9-e9d1-4365-be91-862ebd26bdf4.csv")

# Preview the data
df.head()
```

## ✅ Step 3: Transform - Clean and Prepare the Data

```
# Check for missing values
print("Missing values in each column:\n")
print(df.isnull().sum())
```

```
# Drop columns that may not be useful for analysis (like image paths)
df = df.drop(['Image', 'ImageURL'], axis=1)

# Encode categorical columns using one-hot encoding
df_encoded = pd.get_dummies(df, columns=['Gender', 'Category', 'SubCategory', 'ProductType', 'Colour', 'Usage'])

# Preview transformed data
df_encoded.head()
```

## ✅ Step 4: Load - Save the Cleaned Dataset

```
# Save the transformed dataset to CSV
df_encoded.to_csv("fashion_cleaned.csv", index=False)
print("Transformed dataset saved as 'fashion_cleaned.csv'")
```
